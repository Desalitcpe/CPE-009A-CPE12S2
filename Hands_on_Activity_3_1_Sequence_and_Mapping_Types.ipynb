```python
# Distance Formula Program
# Calculate the distance between two points given a list of coordinates

import math

# Given coordinates list
coordinates_list = [(1, 1), (2, 3)]

# Extract points
point1 = coordinates_list[0]  # (1, 1)
point2 = coordinates_list[1]  # (2, 3)

# Calculate distance using distance formula: d = √[(x₂ - x₁)² + (y₂ - y₁)²]
x1, y1 = point1
x2, y2 = point2

distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# Display the result
print("=" * 50)
print("DISTANCE FORMULA CALCULATOR")
print("=" * 50)
print(f"Point 1: {point1}")
print(f"Point 2: {point2}")
print(f"Distance between points: {distance:.4f}")
print("=" * 50)

# Additional functionality: Calculate distance for multiple pairs
print("\nEXTRA: Calculate distance for custom coordinates")
print("-" * 50)

def calculate_distance(coord_list):
    """Calculate distance between two points in a list"""
    if len(coord_list) >= 2:
        x1, y1 = coord_list[0]
        x2, y2 = coord_list[1]
        return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    else:
        return None

# Test with more examples
test_coordinates = [
    [(0, 0), (3, 4)],      # Should be 5.0
    [(5, 5), (8, 9)],      # Should be 5.0
    [(-1, -1), (2, 3)]     # Should be 5.0
]

for i, coords in enumerate(test_coordinates, 1):
    dist = calculate_distance(coords)
    print(f"Example {i}: {coords[0]} to {coords[1]} = {dist:.4f}")
```

```python
# Simple Word Filter Program
# Replace all occurrences of "stupid" with asterisks (*) equal to the word length

print("=" * 50)
print("SIMPLE WORD FILTER")
print("=" * 50)

# Get input from user
user_input = input("Enter a sentence: ")
print(f"\nOriginal: {user_input}")

# Word to filter
bad_word = "stupid"

# Method 1: Using replace() function
filtered_text = user_input.replace(bad_word, "*" * len(bad_word))
print(f"Filtered: {filtered_text}")

# Method 2: Case-insensitive filtering
print("\n" + "-" * 30)
print("CASE-INSENSITIVE FILTERING")
print("-" * 30)

def filter_stupid(text):
    """Replace 'stupid' in any case with asterisks"""
    words = text.split()
    filtered_words = []

    for word in words:
        # Remove punctuation for checking
        clean_word = word.lower().strip('.,!?;:')

        if clean_word == "stupid":
            # Preserve original punctuation
            if word[-1] in '.,!?;:':
                filtered_words.append("*" * len(bad_word) + word[-1])
            else:
                filtered_words.append("*" * len(bad_word))
        else:
            filtered_words.append(word)

    return " ".join(filtered_words)

# Test with various inputs
test_phrases = [
    "This is stupid",
    "That was STUPID!",
    "Don't be stupid, please.",
    "Stupid stupid stupid",
    "This sentence has no bad words"
]

print("\nTEST EXAMPLES:")
for i, phrase in enumerate(test_phrases, 1):
    result = filter_stupid(phrase)
    print(f"{i}. '{phrase}'")
    print(f"   → '{result}'")

# Interactive filtering
print("\n" + "-" * 30)
print("INTERACTIVE FILTER")
print("-" * 30)

while True:
    text = input("\nEnter text (or 'quit' to exit): ")
    if text.lower() == 'quit':
        break

    # Check for the word in any case
    if 'stupid' in text.lower():
        filtered = filter_stupid(text)
        print(f"Filtered: {filtered}")
    else:
        print("No inappropriate words found!")
```

```python
# Simple Phonebook Program
# Display person's information based on ID input

print("=" * 50)
print("SIMPLE PHONEBOOK DIRECTORY")
print("=" * 50)

# List of dictionaries containing contact information
phonebook = [
    {
        'id': 1,
        'first_name': 'John',
        'last_name': 'Doe',
        'contact': '09171234567',
        'address': '123 Main St, Manila'
    },
    {
        'id': 2,
        'first_name': 'Jane',
        'last_name': 'Smith',
        'contact': '09281234567',
        'address': '456 Oak Ave, Quezon City'
    },
    {
        'id': 3,
        'first_name': 'Robert',
        'last_name': 'Johnson',
        'contact': '09391234567',
        'address': '789 Pine Rd, Makati'
    },
    {
        'id': 4,
        'first_name': 'Maria',
        'last_name': 'Garcia',
        'contact': '09151234567',
        'address': '321 Cedar St, Pasig'
    },
    {
        'id': 5,
        'first_name': 'David',
        'last_name': 'Brown',
        'contact': '09451234567',
        'address': '654 Birch Ln, Taguig'
    }
]

# Display all contacts
print("\nALL CONTACTS:")
print("-" * 50)
print(f"{'ID':<5} {'NAME':<25} {'CONTACT':<15} {'ADDRESS'}")
print("-" * 50)

for contact in phonebook:
    full_name = f"{contact['first_name']} {contact['last_name']}"
    print(f"{contact['id']:<5} {full_name:<25} {contact['contact']:<15} {contact['address']}")

# Function to find contact by ID
def find_contact_by_id(contact_list, search_id):
    """Find and return contact information based on ID"""
    for contact in contact_list:
        if contact['id'] == search_id:
            return contact
    return None

# Function to display contact details
def display_contact(contact):
    """Display formatted contact information"""
    print("\n" + "=" * 40)
    print("CONTACT DETAILS")
    print("=" * 40)
    print(f"ID:          {contact['id']}")
    print(f"First Name:  {contact['first_name']}")
    print(f"Last Name:   {contact['last_name']}")
    print(f"Full Name:   {contact['first_name']} {contact['last_name']}")
    print(f"Contact #:   {contact['contact']}")
    print(f"Address:     {contact['address']}")
    print("=" * 40)

# Main program loop
print("\n" + "=" * 50)
print("PHONEBOOK SEARCH")
print("=" * 50)

while True:
    print("\nOptions:")
    print("1. Search by ID")
    print("2. View all contacts")
    print("3. Exit")

    choice = input("\nEnter your choice (1-3): ").strip()

    if choice == '1':
        try:
            search_id = int(input("Enter ID number to search: "))
            result = find_contact_by_id(phonebook, search_id)

            if result:
                display_contact(result)
            else:
                print(f"\n No contact found with ID: {search_id}")
        except ValueError:
            print("\n Invalid input! Please enter a number.")

    elif choice == '2':
        print("\n" + "-" * 60)
        print(f"{'ID':<5} {'NAME':<25} {'CONTACT':<15} {'ADDRESS'}")
        print("-" * 60)
        for contact in phonebook:
            full_name = f"{contact['first_name']} {contact['last_name']}"
            print(f"{contact['id']:<5} {full_name:<25} {contact['contact']:<15} {contact['address']}")

    elif choice == '3':
        print("\nThank you for using the Phonebook Directory. Goodbye!")
        break

    else:
        print("\n Invalid choice! Please enter 1, 2, or 3.")

# Additional functionality: Search by name
print("\n" + "=" * 50)
print("BONUS: SEARCH BY NAME")
print("=" * 50)

def search_by_name(contact_list, search_name):
    """Search contacts by first or last name"""
    results = []
    search_name = search_name.lower()

    for contact in contact_list:
        if (search_name in contact['first_name'].lower() or
            search_name in contact['last_name'].lower()):
            results.append(contact)

    return results

# Test name search
name_search = input("\nEnter name to search (or press Enter to skip): ")
if name_search:
    matches = search_by_name(phonebook, name_search)

    if matches:
        print(f"\nFound {len(matches)} contact(s):")
        for match in matches:
            print(f"ID: {match['id']} - {match['first_name']} {match['last_name']} - {match['contact']}")
    else:
        print("No matches found.")
```
