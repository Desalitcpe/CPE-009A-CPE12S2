```python
def remainder(n, m):
  while True:
    if n - m < 0:
      return n
    else:
      n = n - m

print(remainder(10, 4))
```

# this is **DICTIONARY!!!!!**

```python
var = "hello world"
vardict = {'first' : 123, 2 : 'abc', '3' : var, 4 : ['lista', 'listb']}
print(vardict['first'])
```
