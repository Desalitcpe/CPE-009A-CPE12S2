Develop a python scripts that:

    Calculate total marks and percentage
    Assign grade for each subject based on marks:

90-100: A+

80-89: A

70-79: B+

60-69: B

50-59: C

40-49: D

Below 40: F (Fail)

    Determine overall result (Pass/Fail)



Performance Remarks:


Percentage ≥ 90: "Outstanding"

Percentage ≥ 80: "Excellent"

Percentage ≥ 70: "Very Good"

Percentage ≥ 60: "Good"

Percentage ≥ 40: "Satisfactory"

Below 40: "Needs Improvement"



```python
def get_performance_remark(percentage):
  if percentage >= 90:
    return "Outstanding"
  elif percentage >= 80:
    return "Excellent"
  elif percentage >= 70:
    return "Very Good"
  elif percentage >= 60:
    return "Good"
  elif percentage >= 40:
    return "Satisfactory"
  else:
    return "Needs Improvement"

def get_subject_remark(sub_percentage):
  if sub_percentage >= 90:
    return "A+"
  elif sub_percentage >= 80:
    return "A"
  elif sub_percentage >= 70:
    return "B+"
  elif sub_percentage >= 60:
    return "B"
  elif sub_percentage >= 50:
    return "C"
  elif sub_percentage >= 40:
    return "D"
  elif sub_percentage < 40:
    return "F"


num_subjects = int(input("Enter the number of subjects: "))

total_marks_obtained = 0
total_possible_marks = 0

for i in range(num_subjects):

    subject_mark = float(input(f"Enter marks obtained for subject {i+1}: "))
    max_subject_mark = float(input(f"Enter maximum possible marks for subject {i+1}: "))


    total_marks_obtained += subject_mark
    total_possible_marks += max_subject_mark


    if max_subject_mark > 0:
        subject_percentage = (subject_mark / max_subject_mark) * 100
        print(f"Subject {i+1} Percentage: {subject_percentage:.2f}%")
        print(f"Subject grading {get_subject_remark(subject_percentage)}")
    else:
        print(f"Cannot calculate percentage for subject {i+1}: Maximum marks is zero.")


if total_possible_marks > 0:
    overall_percentage = (total_marks_obtained / total_possible_marks) * 100
    print(f"\nOverall Total Marks Obtained: {total_marks_obtained:.2f}")
    print(f"Overall Total Possible Marks: {total_possible_marks:.2f}")
    print(f"Overall Percentage: {overall_percentage:.2f}%")
    print(f"Performance Remark: {get_performance_remark(overall_percentage)}")
else:
    print("\nNo subjects entered or total possible marks are zero. Cannot calculate overall percentage.")
```
